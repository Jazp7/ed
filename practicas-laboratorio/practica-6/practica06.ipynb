### 5.1 Requerimiento 1: Mapeo de Registros (Inventario por SKU)
Diseñar una estructura de inventario donde las claves sean códigos SKU y los valores sean diccionarios internos con atributos (`nombre`, `precio`, `existencia`, `categoria`).

```python
# Estructura de Inventario indexada por SKU
inventario_sku = {
    "SKU-1001": {"nombre": "Teclado Mecánico RGB", "precio": 59.99, "existencia": 25, "categoria": "Periféricos"},
    "SKU-1002": {"nombre": "Mouse Óptico Inalámbrico", "precio": 24.50, "existencia": 40, "categoria": "Periféricos"},
    "SKU-1003": {"nombre": "Monitor Gamer 24\" IPS 144Hz", "precio": 175.00, "existencia": 12, "categoria": "Pantallas"},
    "SKU-1004": {"nombre": "Auriculares HyperX Cloud II", "precio": 69.99, "existencia": 18, "categoria": "Audio"},
    "SKU-1005": {"nombre": "Disco SSD NVMe 1TB M.2", "precio": 89.90, "existencia": 30, "categoria": "Almacenamiento"},
    "SKU-1006": {"nombre": "Monitor Curvo 27\" QHD", "precio": 260.00, "existencia": 8, "categoria": "Pantallas"}
}

# Consultas: Acceso directo vs Acceso seguro con .get()
print("--- Consultas de Inventario ---")
print("Búsqueda directa (SKU-1001):", inventario_sku["SKU-1001"]["nombre"])
print("Búsqueda segura (SKU-9999):", inventario_sku.get("SKU-9999", "Producto no encontrado"))

# Actualización y modificación de existencias y precios
print("\n--- Actualización de Stock y Precio ---")
inventario_sku["SKU-1001"]["existencia"] += 10
inventario_sku["SKU-1001"]["precio"] = 54.99
print(f"SKU-1001 actualizado -> Stock: {inventario_sku['SKU-1001']['existencia']} | Precio: ${inventario_sku['SKU-1001']['precio']}")

# Inserción de un nuevo registro SKU
inventario_sku["SKU-1007"] = {
    "nombre": "Micrófono Condensador USB",
    "precio": 42.00,
    "existencia": 15,
    "categoria": "Audio"
}
print(f"Nuevo producto insertado: {inventario_sku['SKU-1007']['nombre']}")

# Eliminación segura con .pop()
eliminado = inventario_sku.pop("SKU-1006", None)
print(f"Producto eliminado: {eliminado['nombre'] if eliminado else 'No encontrado'}")

# Catálogo completo ordenado
print("\n--- Catálogo Actualizado ---")
for sku, datos in inventario_sku.items():
    print(f"SKU: {sku:<10} | {datos['nombre']:<30} | ${datos['precio']:>6.2f} | Stock: {datos['existencia']}")
```

### 5.2 Requerimiento 2: Contador de Ocurrencias (Monitoreo con .get())
Procesar una lista de transacciones o registros de eventos y calcular la frecuencia de cada elemento sin usar librerías externas, aplicando `.get()` con valores por defecto.

```python
# Lista de registros de eventos del sistema (Logs)
registro_eventos = [
    "LOGIN_SUCCESS", "VENTA", "LOGIN_SUCCESS", "FALLO_AUTH",
    "CONSULTA_STOCK", "VENTA", "VENTA", "REPOSICION",
    "CONSULTA_STOCK", "LOGIN_SUCCESS", "FALLO_AUTH", "VENTA"
]

# Lista de transacciones de ventas por SKU
transacciones_sku = [
    "SKU-1001", "SKU-1002", "SKU-1001", "SKU-1005", "SKU-1001",
    "SKU-1003", "SKU-1002", "SKU-1001", "SKU-1005", "SKU-1004"
]

# Conteo de ocurrencias aplicando .get(clave, 0) + 1
frecuencia_eventos = {}
for evento in registro_eventos:
    frecuencia_eventos[evento] = frecuencia_eventos.get(evento, 0) + 1

frecuencia_ventas = {}
for sku in transacciones_sku:
    frecuencia_ventas[sku] = frecuencia_ventas.get(sku, 0) + 1

print("--- Frecuencia de Eventos del Sistema ---")
for ev, freq in frecuencia_eventos.items():
    barra = "#" * freq
    print(f"Evento: {ev:<18} | Ocurrencias: {freq:>2} | {barra}")

print("\n--- Frecuencia de Ventas por SKU ---")
for sku, ventas in frecuencia_ventas.items():
    nombre = inventario_sku.get(sku, {}).get("nombre", "Desconocido")
    print(f"SKU: {sku:<10} | {nombre:<28} | Ventas: {ventas}")
```

### 5.3 Requerimiento 3: Agrupamiento por Categoría
Organizar una lista de elementos en un diccionario donde cada clave sea una categoría y su valor asociado sea una lista con los elementos pertenecientes a ella.

```python
# Lista de productos heterogéneos con categoría
catalogo_productos = [
    {"sku": "SKU-1001", "nombre": "Teclado Mecánico RGB", "categoria": "Periféricos"},
    {"sku": "SKU-1002", "nombre": "Mouse Óptico Inalámbrico", "categoria": "Periféricos"},
    {"sku": "SKU-1003", "nombre": "Monitor Gamer 24\" IPS 144Hz", "categoria": "Pantallas"},
    {"sku": "SKU-1004", "nombre": "Auriculares HyperX Cloud II", "categoria": "Audio"},
    {"sku": "SKU-1005", "nombre": "Disco SSD NVMe 1TB M.2", "categoria": "Almacenamiento"},
    {"sku": "SKU-1007", "nombre": "Micrófono Condensador USB", "categoria": "Audio"},
    {"sku": "SKU-1008", "nombre": "Memoria RAM DDR5 32GB", "categoria": "Almacenamiento"},
    {"sku": "SKU-1009", "nombre": "Webcam 4K Streaming Pro", "categoria": "Periféricos"}
]

# Agrupamiento por categoría en un diccionario de listas
productos_por_categoria = {}
for prod in catalogo_productos:
    cat = prod["categoria"]
    if cat not in productos_por_categoria:
        productos_por_categoria[cat] = []
    productos_por_categoria[cat].append(prod["nombre"])

print("--- Productos Agrupados por Categoría ---")
for categoria, items in productos_por_categoria.items():
    print(f"\nCategoría: {categoria} ({len(items)} productos)")
    for item in items:
        print(f"  * {item}")

# Agregar dinámicamente un nuevo producto a una categoría existente
productos_por_categoria["Periféricos"].append("Pad Mouse Gamer XXL")
print("\nPeriféricos actualizados tras agregar producto:")
print(productos_por_categoria["Periféricos"])
```

### 5.4 Sistema Integrado: Monitoreo e Inventario Hash
Estructuración de una clase completa `SistemaMonitoreoInventarioHash` que unifica los tres conceptos en un flujo operativo real con bitácora de eventos, control de existencias y reportes estructurados.

```python
class SistemaMonitoreoInventarioHash:
    def __init__(self):
        self.inventario = {}
        self.logs_eventos = []

    def registrar_producto(self, sku, nombre, precio, existencia, categoria):
        self.inventario[sku] = {
            "nombre": nombre,
            "precio": precio,
            "existencia": existencia,
            "categoria": categoria
        }
        self.logs_eventos.append(f"ALTA:{sku}")

    def vender(self, sku, cantidad=1):
        producto = self.inventario.get(sku)
        if not producto:
            self.logs_eventos.append("ERROR_SKU_NO_EXISTE")
            print(f"❌ Error: SKU '{sku}' no existe.")
            return False
        if producto["existencia"] < cantidad:
            self.logs_eventos.append("ERROR_STOCK_INSUFICIENTE")
            print(f"⚠️ Stock insuficiente para {producto['nombre']}. Disponibles: {producto['existencia']}")
            return False
        producto["existencia"] -= cantidad
        self.logs_eventos.append("VENTA_EXITOSA")
        self.logs_eventos.append(f"VENTA:{sku}")
        print(f"✅ Venta realizada: {cantidad}x {producto['nombre']} | Stock restante: {producto['existencia']}")
        return True

    def metricas_frecuencia(self):
        conteo = {}
        for ev in self.logs_eventos:
            conteo[ev] = conteo.get(ev, 0) + 1
        return conteo

    def inventario_por_categoria(self):
        agrupados = {}
        for sku, datos in self.inventario.items():
            cat = datos["categoria"]
            if cat not in agrupados:
                agrupados[cat] = []
            agrupados[cat].append(f"{sku} - {datos['nombre']} (${datos['precio']:.2f})")
        return agrupados

# Demostración del sistema integrado
sistema = SistemaMonitoreoInventarioHash()
sistema.registrar_producto("SKU-A1", "Cámara Web HD", 29.99, 10, "Video")
sistema.registrar_producto("SKU-A2", "Foco LED RGB Inteligente", 12.50, 25, "Iluminación")
sistema.registrar_producto("SKU-A3", "Tira LED 5 Metros", 18.00, 15, "Iluminación")

sistema.vender("SKU-A1", 2)
sistema.vender("SKU-A2", 3)
sistema.vender("SKU-A999", 1)

print("\n--- Frecuencia de Eventos Registrados (.get) ---")
for ev, count in sistema.metricas_frecuencia().items():
    print(f"  {ev:<25} -> {count}")

print("\n--- Catálogo Agrupado por Categorías ---")
for cat, prods in sistema.inventario_por_categoria().items():
    print(f"  [{cat}]")
    for p in prods:
        print(f"    - {p}")
```
